```python
%sql
CREATE OR REPLACE TABLE ecommerce.gold.gold_customer_segments AS

SELECT
    customer_segment,
    segment_category,

    COUNT(*) AS customer_count,

    ROUND(AVG(age), 2) AS avg_age,

    ROUND(AVG(total_purchases), 2) AS avg_purchases,

    ROUND(AVG(avg_order_value_usd), 2) AS avg_order_value_usd,

    ROUND(SUM(total_spent_usd), 2) AS total_revenue_usd,

    ROUND(AVG(customer_lifetime_value_usd), 2) 
        AS avg_customer_lifetime_value_usd,

    ROUND(AVG(customer_profitability_usd), 2) 
        AS avg_customer_profitability_usd,

    ROUND(AVG(satisfaction_score), 2) 
        AS avg_satisfaction_score,

    ROUND(AVG(churn_risk_score), 2) 
        AS avg_churn_risk_score,

    ROUND(AVG(rfm_score), 2) 
        AS avg_rfm_score

FROM ecommerce.silver.silver_customer_data

GROUP BY
    customer_segment,
    segment_category;
```
